```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import kagglehub as kh
from os import environ
%matplotlib inline
```

Studying the dataset.

```python
environ["KAGGLEHUB_ANONYMOUS_DOWNLOAD"]='1'
path=kh.dataset_download("nelgiriyewithana/mcdonalds-store-reviews")
df=pd.read_csv(f"{path}/McDonald_s_Reviews.csv", encoding='utf-8', encoding_errors='ignore')
df.info(memory_usage='deep')
print(df.shape)
print(df.describe())
```

Output the first 10 rows of the dataset.

```python
display(df.head(10))
```

Removing the limit on the maximum length of text in cells when displaying tables on the screen. Displaying 10 rows of the store_address and review columns from the dataset.

```python
pd.set_option('display.max_colwidth', None)
display(df[['store_address', 'review']].head(10))
```

Output the unique count of brands and restaurant types. Output the count of NaN elements and duplicates.

```python
print(df.groupby('store_name')['store_address'].nunique())
print(df.groupby('category')['store_address'].nunique())
print(df.isna().sum())
print(df.duplicated().sum())
```

Removing duplicates and the store_name and category columns from the dataset. Converting the rating_count and rating columns to numeric values. Outputting the first 10 rows of the modified dataset.

```python
df=df.dropna().drop(['store_name', 'category'], axis=1)
df['reviewer_id']=df['reviewer_id'].astype('int32')
df['rating_count']=df['rating_count'].str.replace(',', '').astype('int32')
df['rating']=df['rating'].str.split().str[0].astype('int32')
display(df.head(10))
```

Creating a function to convert a relative date in a string value to an extended date format. Converting the review_date column to the datetime type using the created function. Outputting the first 10 rows of the modified dataset.

```python
def make_date(value):
    if value.startswith('a'):
        value=value.replace('a', '1', 1)
    num, word=value.split()[:2]
    num=int(num)
    if word in ['hour', 'hours']:
        offset=pd.DateOffset(hours=num)
    elif word in ['day', 'days']:
        offset=pd.DateOffset(days=num)
    elif word in ['week', 'weeks']:
        offset=pd.DateOffset(weeks=num)
    elif word in ['month', 'months']:
        offset=pd.DateOffset(months=num)
    elif word in ['year', 'years']:
        offset=pd.DateOffset(years=num)
    else:
        return np.nan
    result=pd.Timestamp('2023-06-19')-offset
    return result.strftime('%Y-%m-%d')
df['review_time']=pd.to_datetime(df['review_time'].apply(make_date))
display(df.head(10))
```

Checking the contents of the modified dataset.

```python
df.info(memory_usage='deep')
print(df.shape)
print(df.describe())
```

Check for duplicates and NaN elements.

```python
print(df.isna().sum())
print(df.duplicated().sum())
```

Output the unique count of restaurants and countries, and list all unique restaurants.

```python
print(df['store_address'].nunique())
print(df['store_address'].str.endswith('United States').nunique())
print(df['store_address'].unique().tolist())
```

Correction of anomalies in two cells of the store_address column.

```python
df.loc[
    df['store_address']=="151 West 34th Street (Macy's 7th Floor, New York, NY 10001, United States",
    'store_address']="151 West 34th Street (Macy's 7th Floor), New York, NY 10001, United States"
df.loc[
    df['store_address']=="By Mandalay Bay, 3999 S Las Vegas Blvd, Las Vegas, NV 89119, United States",
    'store_address']="By Mandalay Bay 3999 S Las Vegas Blvd, Las Vegas, NV 89119, United States"
```

Creating new columns (street, city, state, zip_code) based on the store_address column. Removing the store_address column from the dataset. Changing the order of columns in the dataset. Displaying the first 10 rows of the modified dataset.

```python
lst1=df['store_address'].str.split(', ')
df['street']=lst1.str[0]
df['city']=lst1.str[1]
df[['state', 'zip_code']]=lst1.str[2].str.split(' ', expand=True)
df=df.drop('store_address', axis=1)
df=df[['reviewer_id', 'street', 'city', 'state', 'zip_code', 'latitude ', 'longitude', 'rating_count', 'review_time', 'rating', 'review']]
display(df.head(10))
```

Checking the contents of the modified dataset.

```python
df.info(memory_usage='deep')
print(df.shape)
```

Check for duplicates and NaN elements.

```python
print(df.duplicated().sum())
print(df.isna().sum())
```

Removing elements '�' from the cells of the review column. Changing the indices in the dataset. Outputting the first 10 rows of the modified dataset.

```python
df['review']=df['review'].str.replace('�', '')
df=df.rename(columns={'latitude ': 'latitude'})
df.index=np.arange(0, len(df))
display(df.head(10))
```

Output of the first 10 rows of the dataset without the review column, sorted by the rating and rating_count columns in descending order.

```python
display(df.iloc[:, :10].sort_values(by=['rating', 'rating_count'], ascending=False).head(10))
```

Creating the state_rating table based on grouping by the state column. Including in the new table columns for the unique number of streets, the average rounded rating, and the sum of the reviews left. Outputting the new table.

```python
state_rating=df.groupby('state')[['street', 'rating', 'rating_count']].agg(
    {'street': 'nunique', 'rating': 'mean', 'rating_count': 'sum'})
state_rating=state_rating.reset_index()
state_rating['rating']=np.round(state_rating['rating'], 1)
display(state_rating)
```

Creating a bar chart of the state_rating table.

```python
ax=state_rating.sort_values(by=['rating', 'rating_count'], ascending=False)[::-1].plot(
    kind='barh', x='state', y='rating', color='blue', figsize=(11, 6))
for container in ax.containers:
    ax.bar_label(container, padding=2)
plt.title('Rating of States', fontsize=15, fontweight='bold')
plt.xlabel('Rating', fontsize=10, color='grey')
plt.ylabel('States', fontsize=10, color='grey')
plt.show()
```

Creating the city_rating table based on grouping by the city column. Including in the new table columns for the unique number of streets, the average rounded rating, and the sum of reviews left. Outputting the new table.

```python
city_rating=df.groupby('city')[['street', 'rating', 'rating_count']].agg(
    {'street': 'nunique', 'rating': 'mean', 'rating_count': 'sum'})
city_rating=city_rating.reset_index()
city_rating['rating']=np.round(city_rating['rating'], 1)
display(city_rating)
```

Creating a bar chart of the city_rating table.

```python
ax=city_rating.sort_values(by=['rating', 'rating_count'], ascending=False)[::-1].plot(
    kind='barh', x='city', y='rating', color='orange', figsize=(11, 6))
for container in ax.containers:
    ax.bar_label(container, padding=2)
plt.title('Rating of Cities', fontsize=15, fontweight='bold')
plt.xlabel('Rating', fontsize=10, color='grey')
plt.ylabel('Cities', fontsize=10, color='grey')
plt.show()
```

Creating a new table called negative_reviews based on the content of negative words in the cells of the review dataset column. Outputting the first five rows of the new table.

```python
negative_words='|'.join(["wait", "waiting", "slow", "line", "queue", "forever", "delayed",
                  "cold", "wrong", "missing", "stale", "old", "dry", "burnt", "taste",
                  "rude", "unfriendly", "ignored", "attitude", "terrible", "worst", "horrible",
                  "dirty", "gross", "nasty", "mess", "cleanliness",
                  "disappointed", "waste", "never", "bad", "avoid", "sucks"])
negative_reviews=df[df['review'].str.contains(negative_words, case=False, na=False)]
display(negative_reviews.head())
```

Creating a new table called antirating based on grouping the city and street columns. Including the number of reviews, the average rounded rating, and the restaurant’s location in the new table. Outputting the new table sorted in descending order by review and in ascending order by rating.

```python
antirating=negative_reviews.groupby(['city', 'street'], as_index=False)[['review', 'rating']].agg(
    {'review': 'count', 'rating': 'mean'})
antirating['rating']=np.round(antirating['rating'], 1)
antirating['city']=antirating['street'].str.cat(antirating['city'], sep=', ')
antirating=antirating.drop('street', axis=1).rename(columns={'city': 'address'})
display(antirating.sort_values(by=['review', 'rating'], ascending=[False, True]))
```

Creating a bar chart of the antirating table.

```python
antirating1=antirating.sort_values(by=['review', 'rating'], ascending=[False, True]).head(20).iloc[::-1]
ax=antirating1.plot(
    kind='barh', x='address', y='review', color='red', figsize=(11, 6))
for container in ax.containers:
    ax.bar_label(container, padding=2)
plt.title('Top 20 worst McDonald’s restaurants', fontsize=15, fontweight='bold')
plt.xlabel('Count of Negative Reviews', fontsize=10, color='grey')
plt.ylabel('Restaurant addresses', fontsize=10, color='grey')
plt.show()
```

Output: the number of states, the number of cities, the number of restaurants, the average rating for all restaurants, the number of positive reviews, and the number of negative reviews.

```python
info=pd.Series({
    'count_state': df['state'].nunique(),
    'count_city': df['city'].nunique(),
    'count_restaurants': df['street'].nunique(),
    'average_rating': df['rating'].mean(),
    'count_positive_review': df['review'].count()-negative_reviews['review'].count(),
    'count_negative_review': negative_reviews['review'].count()
}).astype('int32')
display(info)
```
